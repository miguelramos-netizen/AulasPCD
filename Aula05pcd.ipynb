```python
import csv
import json
import random

# Gerar 20 números aleatórios entre 1 e 100
numeros = [random.randint(1, 100) for _ in range(20)]

# Preparar a estrutura de dados com os três campos
dados = []
for num in numeros:
  dados.append(
      {"valor_original": num, "valor_x2": num * 2, "valor_x10": num * 10}
  )

# 1. Salvar em TXT
txt_filename = "resultados_numeros.txt"
with open(txt_filename, "w", encoding="utf-8") as txt_file:
  txt_file.write("Valor Original | Valor * 2 | Valor * 10\n")
  txt_file.write("-" * 40 + "\n")
  for item in dados:
    txt_file.write(
        f"{item['valor_original']:<14} | {item['valor_x2']:<9} |"
        f" {item['valor_x10']:<10}\n"
    )

# 2. Salvar em JSON
json_filename = "resultados_numeros.json"
with open(json_filename, "w", encoding="utf-8") as json_file:
  json.dump(dados, json_file, indent=4, ensure_ascii=False)

# 3. Salvar em CSV
csv_filename = "resultados_numeros.csv"
with open(csv_filename, "w", newline="", encoding="utf-8") as csv_file:
  fieldnames = ["valor_original", "valor_x2", "valor_x10"]
  writer = csv.DictWriter(csv_file, fieldnames=fieldnames)

  writer.writeheader()
  writer.writerows(dados)
```
