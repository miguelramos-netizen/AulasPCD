```python
import pandas as pd
#Importando um arquivo .csv dos dados de temperatura dos municípios
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
santa_rosa = df[(df["nome_municipio"] == "Santa Rosa") & (df["uf"] == "RS")]
santa_rosa
```

```python
colunas_tmin = [c for c in df.columns if c.startswith("tmin_")]
# Pega o nome da coluna com o menor valor e extrai a sigla do mês
mes_temp_min = (
    santa_rosa[colunas_tmin].idxmin(axis=1).values[0].replace("tmin_", "")
)
val_temp_min = santa_rosa[colunas_tmin].min(axis=1).values[0]

print(f"\nMês de menor temperatura mínima: {mes_temp_min} ({val_temp_min}°C)")
```

```python
colunas_def = [c for c in df.columns if c.startswith("def_")]
# Pega o nome da coluna com o maior valor e extrai a sigla do mês
mes_max_def = (
    santa_rosa[colunas_def].idxmax(axis=1).values[0].replace("def_", "")
)
val_max_def = santa_rosa[colunas_def].max(axis=1).values[0]

print(f"Mês de maior déficit hídrico: {mes_max_def} ({val_max_def} mm/mês)")
```

```python
# MÉDIA DE TEMPERATURA PELAS 4 ESTAÇÕES DO ANO (Hemisfério Sul)
verao = santa_rosa[
    ["temp_media_dez", "temp_media_jan", "temp_media_fev"]
].mean(axis=1).values[0]
outono = santa_rosa[
    ["temp_media_mar", "temp_media_abr", "temp_media_mai"]
].mean(axis=1).values[0]
inverno = santa_rosa[
    ["temp_media_jun", "temp_media_jul", "temp_media_ago"]
].mean(axis=1).values[0]
primavera = santa_rosa[
    ["temp_media_set", "temp_media_out", "temp_media_nov"]
].mean(axis=1).values[0]

print("\n--- MÉDIA DE TEMPERATURA POR ESTAÇÃO (°C) ---")
print(f"Verão: {verao:.2f}°C")
print(f"Outono: {outono:.2f}°C")
print(f"Inverno: {inverno:.2f}°C")
print(f"Primavera: {primavera:.2f}°C")
```
