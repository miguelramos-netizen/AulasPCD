Construa uma calculadora básica usando os conceitos de orientação a objetos.
Use a função raiz quadrada usando o método de Newton (se achar complicado use a função SQRT).

```python
import math
class calc:
    # método construtor
    def __init__(self, n1: float, n2: float, tipo:str):
        self.n1 = n1
        self.n2 = n2
        self.__tipo = tipo

    def setn1(self, n1: float):
        self.n1 = n1

    def getn1(self):
        return self.n1

    def setn2(self, n2: float):
        self.n2 = n2

    def getn2(self):
        return self.n2

    def setTipo(self, tipo: str):
        self.__tipo = tipo

    def getTipo(self):
        return self.__tipo



    def soma(self):
        print(f'Soma de {self.n1} e {self.n2} é {self.n1 + self.n2}')

    def subtracao(self):
        print(f'Subtração de {self.n1} e {self.n2} é {self.n1 - self.n2}')

    def multiplicacao(self):
        print(f'Multiplicação de {self.n1} e {self.n2} é {self.n1 * self.n2}')

    def divisao(self):
        print(f'Divisão de {self.n1} e {self.n2} é {self.n1 / self.n2}')


    def raiz(self):
        print(f'Raiz Quadrada de {self.n1} é {math.sqrt(self.n1)}')

    def escolhern1(self):
        self.n1 = float(input("Digite o primeiro número  "))

    def escolhern2(self):
        self.n2 = float(input("Digite o segundo número  "))




teste = calc(n1 = -1, n2 = -1, tipo = "cacl")
teste.escolhern1()
teste.escolhern2()

xx = int(input("Escolha a operação com esses números 1 = soma, 2 = subtração, 3 = multiplicação, 4 = divisão, 5 = raiz quadrada do número 1 -  "))

if xx == 1:
  teste.soma()
elif xx == 2:
  teste.subtracao()
elif xx == 3:
  teste.multiplicacao()
elif xx == 4:
  teste.divisao()
elif xx == 5:
  teste.raiz()
else:
  print("você não digitou um número de 1 à 5")


```
